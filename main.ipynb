# Step function

```python
def step_function(x):
    return 1 * (x > 0)
```

```python
step_function(-10), step_function(10)
```

# Sigmoid

```python
def sigmoid_function(x):
    return 1 / (1 + np.exp(-x))
```

```python
sigmoid_function(-10), sigmoid_function(10), sigmoid_function(1)
```

# Hyperbolic tangent

```python
def hyperbolic_tangent_function(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))
```

```python
hyperbolic_tangent_function(-10), hyperbolic_tangent_function(10), hyperbolic_tangent_function(1)
```

# Relu

```python
def relu_function(x):
    return np.maximum(0, x)
```

```python
relu_function(-10), relu_function(10), relu_function(1)
```

# Softmax

```python
def softmax_function(x):
    return np.exp(x) / np.sum(np.exp(x))
```

```python
softmax_function([1, 2, 3])
```

# Linear

```python
def linear_function(x):
    return x
```

```python
linear_function(-10), linear_function(10), linear_function(1)
```
